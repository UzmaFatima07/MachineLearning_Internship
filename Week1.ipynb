# Task 1: Environment Setup & Dataset Loading

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris

# Load the Ames Housing dataset
df = pd.read_csv("train.csv")

# Verify shape
print("Shape:", df.shape)

print("\nFirst 3 rows:")
df.head(3)
```

## Task 2: NumPy Fundamentals

```python
# Create 20 random floats between 0 and 100
arr = np.random.uniform(0, 100, 20)
print("Array:", arr)

# Compute statistics
print("Mean:", np.mean(arr))
print("Median:", np.median(arr))
print("Std Dev:", np.std(arr))
print("Min:", np.min(arr))
print("Max:", np.max(arr))
```

```python
# Create a 5x5 matrix of random integers (0 to 50)
matrix = np.random.randint(0, 50, (5, 5))
print("Matrix:\n", matrix)

# Row-wise sum (axis=1 means along columns, giving one value per row)
print("Row-wise sum:", matrix.sum(axis=1))

# Column-wise mean (axis=0 means along rows, giving one value per column)
print("Column-wise mean:", matrix.mean(axis=0))
```

```python
# Slicing means "picking a portion" of an array
# Example 1: First 5 elements
print("First 5 elements:", arr[:5])

# Example 2: Last 5 elements
print("Last 5 elements:", arr[-5:])

# Example 3: Every other element (step of 2)
print("Every other element:", arr[::2])
```

```python
# Boolean indexing = filter using a True/False condition
# This creates a "mask" — True where value > 50, False otherwise
mask = arr > 50
print("Values greater than 50:", arr[mask])

# You can also write it in one line:
print("Same result:", arr[arr > 50])
```

## Task 3: Pandas Data Exploration

```python
# Load the Iris dataset from scikit-learn
iris = load_iris()

# Convert to a Pandas DataFrame
iris_df = pd.DataFrame(iris.data, columns=iris.feature_names)

# Add the target (class) column — 0, 1, 2 represent 3 flower species
iris_df['target'] = iris.target

# Show first 10 rows
print("First 10 rows:")
iris_df.head(10)
```

```python
# Check data types of each column
print("Data types:")
print(iris_df.dtypes)

# Check shape
print("\nShape:", iris_df.shape)  # Should be (150, 5)

# Check for missing values
print("\nMissing values:")
print(iris_df.isnull().sum())  # All should be 0
```

```python
# .describe() gives count, mean, std, min, max, and quartiles
print("Descriptive Statistics:")
iris_df.describe()
```

```python
# Count how many of each species
print("Samples per class:")
print(iris_df['target'].value_counts())
# 0 = setosa, 1 = versicolor, 2 = virginica (50 each)

# Rename for clarity (optional but nice!)
iris_df['species'] = iris_df['target'].map(
    {0: 'setosa', 1: 'versicolor', 2: 'virginica'}
)

# Filter: only rows where sepal length > 5.5
# The column name from sklearn is 'sepal length (cm)'
col = 'sepal length (cm)'
filtered_df = iris_df[iris_df[col] > 5.5]
print(f"\nRows where sepal length > 5.5: {len(filtered_df)}")
filtered_df.head()
```

 ## Task 4: Visualizations

```python
plt.figure(figsize=(8, 5))
plt.hist(iris_df['sepal length (cm)'], bins=15, color='steelblue', edgecolor='white')
plt.title('Distribution of Sepal Length')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Number of Flowers')
plt.tight_layout()
plt.show()
```

```python
colors = {0: 'tomato', 1: 'steelblue', 2: 'seagreen'}
labels = {0: 'Setosa', 1: 'Versicolor', 2: 'Virginica'}

plt.figure(figsize=(8, 5))
for cls in iris_df['target'].unique():
    subset = iris_df[iris_df['target'] == cls]
    plt.scatter(subset['sepal length (cm)'],
                subset['petal length (cm)'],
                color=colors[cls],
                label=labels[cls],
                alpha=0.7)

plt.title('Sepal Length vs Petal Length by Species')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Petal Length (cm)')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
features = ['sepal length (cm)', 'sepal width (cm)',
            'petal length (cm)', 'petal width (cm)']

plt.figure(figsize=(9, 5))
plt.boxplot([iris_df[col] for col in features],
            tick_labels=['Sepal\nLength', 'Sepal\nWidth',
                    'Petal\nLength', 'Petal\nWidth'])
plt.title('Distribution of All Four Iris Features')
plt.ylabel('Measurement (cm)')
plt.tight_layout()
plt.show()
```
